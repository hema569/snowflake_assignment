```
%%sql -r dataframe_1
-- QUESTION 3: DATA LOADING

USE DATABASE SNOWFLAKE_ASSIGNMENT;
USE SCHEMA TUTORIAL;
USE WAREHOUSE TUTORIAL_WH;

-- Check the uploaded file in the stage
LIST @STUDENT_STAGE;

-- Load the CSV file into the STUDENTS table
COPY INTO STUDENTS
FROM @STUDENT_STAGE
FILE_FORMAT = (
    TYPE = CSV
    SKIP_HEADER = 1
);

-- Verify the loaded data
SELECT *
FROM STUDENTS
ORDER BY STUDENT_ID;

-- Verify total number of records
SELECT COUNT(*) AS TOTAL_RECORDS
FROM STUDENTS;
```

```
%%sql -r dataframe_1
-- =====================================================
-- QUESTION 4: SNOWFLAKE TIME TRAVEL
-- =====================================================

USE DATABASE SNOWFLAKE_ASSIGNMENT;
USE SCHEMA TUTORIAL;
USE WAREHOUSE TUTORIAL_WH;


-- Create Employees table

CREATE OR REPLACE TABLE EMPLOYEES (
    EMP_ID INT,
    EMP_NAME VARCHAR(50),
    DEPARTMENT VARCHAR(50),
    SALARY INT
);


-- Insert records

INSERT INTO EMPLOYEES VALUES
(1, 'Alice', 'IT', 50000),
(2, 'Bob', 'HR', 45000),
(3, 'Charlie', 'Finance', 55000),
(4, 'David', 'IT', 60000);


-- Display original data

SELECT * FROM EMPLOYEES
ORDER BY EMP_ID;


-- Update David's salary

UPDATE EMPLOYEES
SET SALARY = 65000
WHERE EMP_ID = 4;


-- Display updated data

SELECT * FROM EMPLOYEES
ORDER BY EMP_ID;


-- Save the time immediately before deleting Bob

SET BEFORE_DELETE = CURRENT_TIMESTAMP();


-- Delete Bob

DELETE FROM EMPLOYEES
WHERE EMP_ID = 2;


-- Display current table after deletion

SELECT * FROM EMPLOYEES
ORDER BY EMP_ID;


-- Use Time Travel to see the table
-- as it existed before Bob was deleted

SELECT *
FROM EMPLOYEES
BEFORE (
    TIMESTAMP => $BEFORE_DELETE
)
ORDER BY EMP_ID;
```
